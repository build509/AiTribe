```python
import pandas as pd

# Load the CSV into a Pandas DataFrame
df = pd.read_csv(
    'Etherscan-export.csv',
    dtype={
        'Value_OUT(ETH)': pd.Float32Dtype(),
        'Value_IN(ETH)': pd.Float32Dtype(),
        }
)

df['Datetime'] = pd.to_datetime(df['UnixTimestamp'], unit='s')
# Rename 'CurrentValue @...' column so the name does not contain the current Ethereum price
for column in df.columns:
  if 'CurrentValue' in column:
    df.rename(columns={column: 'CurrentValue'}, inplace=True)


# Display the first few rows of the DataFrame
print(df.head())
```

```python
# Investigating failed tx and tx with high gas
# 1. Focus on transaction fees (gas costs)
# Calculate statistics for gas fees
gas_mean = df['TxnFee(ETH)'].mean()
gas_median = df['TxnFee(ETH)'].median()
gas_std = df['TxnFee(ETH)'].std()
gas_max = df['TxnFee(ETH)'].max()

print("=== TRANSACTION FEE ANALYSIS ===")
print(f"Mean gas fee: {gas_mean:.8f} ETH")
print(f"Median gas fee: {gas_median:.8f} ETH")
print(f"Max gas fee: {gas_max:.8f} ETH")

# Identify high gas transactions (e.g., top 5%)
high_gas_threshold = df['TxnFee(ETH)'].quantile(0.95)
high_gas_txs = df[df['TxnFee(ETH)'] > high_gas_threshold]
print(f"High gas transactions (>{high_gas_threshold:.8f} ETH): {len(high_gas_txs)} ({len(high_gas_txs)/len(df)*100:.2f}%)")

# 2. Analyze error codes
if 'ErrCode' in df.columns and not df['ErrCode'].isna().all():
    # Count transactions with error codes
    failed_txs = df[~df['ErrCode'].isna()]
    print("\n=== ERROR CODE ANALYSIS ===")
    print(f"Failed transactions: {len(failed_txs)} ({len(failed_txs)/len(df)*100:.2f}%)")

    # Show distribution of error codes
    error_counts = df['ErrCode'].value_counts(dropna=False)
    print("\nError code distribution:")
    for error, count in error_counts.items():
        error_str = 'No Error (Success)' if pd.isna(error) else error
        percentage = count / len(df) * 100
        print(f"{error_str}: {count} transactions ({percentage:.2f}%)")

    # Analyze failed transactions with high gas fees (potential attack attempts)
    failed_high_gas = df[(~df['ErrCode'].isna()) & (df['TxnFee(ETH)'] > high_gas_threshold)]
    print(f"\nFailed transactions with high gas fees: {len(failed_high_gas)}")
    if len(failed_high_gas) > 0:
        print("Example failed high-gas transactions:")
        cols_to_show = ['Transaction Hash', 'From', 'To', 'TxnFee(ETH)', 'ErrCode', 'Method']
        print(failed_high_gas[cols_to_show].head())

# 3. Address analysis for failed transactions
if 'ErrCode' in df.columns and not df['ErrCode'].isna().all():
    failed_txs = df[~df['ErrCode'].isna()]

    # Addresses with the most failed transactions
    failed_from_counts = failed_txs['From'].value_counts()
    failed_to_counts = failed_txs['To'].value_counts()

    print("\n=== ADDRESSES WITH MOST FAILED TRANSACTIONS ===")
    print("Top senders with failed transactions:")
    for addr, count in failed_from_counts.head(10).items():
        percentage = count / len(failed_txs) * 100
        print(f"{addr}: {count} failed transactions ({percentage:.2f}%)")

    print("\nTop recipients with failed transactions:")
    for addr, count in failed_to_counts.head(10).items():
        percentage = count / len(failed_txs) * 100
        print(f"{addr}: {count} failed transactions ({percentage:.2f}%)")

    # Calculate failure rates per address
    # For addresses with at least 5 transactions
    from_total = df['From'].value_counts()
    from_failure_rates = pd.DataFrame({
        'total_txs': from_total,
        'failed_txs': failed_from_counts
    }).fillna(0)
    from_failure_rates['failure_rate'] = from_failure_rates['failed_txs'] / from_failure_rates['total_txs']

    # Filter addresses with at least 5 transactions
    active_addresses = from_failure_rates[from_failure_rates['total_txs'] >= 5]
    high_failure_addresses = active_addresses.sort_values('failure_rate', ascending=False)

    print("\n=== ADDRESSES WITH HIGHEST FAILURE RATES (min 5 transactions) ===")
    for addr, row in high_failure_addresses.head(10).iterrows():
        print(f"{addr}: {row['failed_txs']}/{row['total_txs']} failed ({row['failure_rate']*100:.2f}%)")

# 4. Methods used in failed transactions
if 'ErrCode' in df.columns and not df['ErrCode'].isna().all() and 'Method' in df.columns:
    failed_txs = df[~df['ErrCode'].isna()]
    failed_methods = failed_txs['Method'].value_counts()

    print("\n=== METHODS WITH MOST FAILURES ===")
    for method, count in failed_methods.head(10).items():
        percentage = count / len(failed_txs) * 100
        print(f"{method}: {count} failures ({percentage:.2f}%)")

    # Calculate failure rates per method
    method_total = df['Method'].value_counts()
    method_failure_rates = pd.DataFrame({
        'total_calls': method_total,
        'failed_calls': failed_methods
    }).fillna(0)
    method_failure_rates['failure_rate'] = method_failure_rates['failed_calls'] / method_failure_rates['total_calls']

    # Filter methods with at least 5 calls
    common_methods = method_failure_rates[method_failure_rates['total_calls'] >= 5]
    high_failure_methods = common_methods.sort_values('failure_rate', ascending=False)

    print("\n=== METHODS WITH HIGHEST FAILURE RATES (min 5 calls) ===")
    for method, row in high_failure_methods.head(10).iterrows():
        print(f"{method}: {row['failed_calls']}/{row['total_calls']} failed ({row['failure_rate']*100:.2f}%)")

# 5. Gas usage by method
if 'Method' in df.columns:
    method_gas = df.groupby('Method')['TxnFee(ETH)'].agg(['mean', 'median', 'std', 'max', 'count'])
    method_gas = method_gas.sort_values('mean', ascending=False)

    print("\n=== GAS USAGE BY METHOD ===")
    print("Methods with highest average gas usage:")

    # Show top 10 methods by average gas usage (with at least 5 transactions)
    high_gas_methods = method_gas[method_gas['count'] >= 5].head(10)
    for method, row in high_gas_methods.iterrows():
        print(f"{method}: {row['mean']:.8f} ETH avg ({row['count']} calls)")

# 6. Identify potential protocol exploits through unusual patterns
# Look for outlier transactions based on gas used relative to the method's normal usage
outlier_txs = []

if 'Method' in df.columns:
    # Calculate per-method gas statistics
    method_groups = df.groupby('Method')

    for method, group in method_groups:
        if len(group) >= 5:  # Only analyze methods with sufficient samples
            method_mean = group['TxnFee(ETH)'].mean()
            method_std = group['TxnFee(ETH)'].std()

            # Find transactions with gas usage > 3 standard deviations from the mean
            threshold = method_mean + 3 * method_std
            outliers = group[group['TxnFee(ETH)'] > threshold]

            for _, outlier in outliers.iterrows():
                outlier_txs.append({
                    'tx_hash': outlier['Transaction Hash'],
                    'method': method,
                    'gas_fee': outlier['TxnFee(ETH)'],
                    'method_avg_gas': method_mean,
                    'deviation': (outlier['TxnFee(ETH)'] - method_mean) / method_std,
                    'from': outlier['From'],
                    'to': outlier['To'],
                    'error_code': outlier['ErrCode'] if 'ErrCode' in outlier else None
                })

print("\n=== POTENTIAL EXPLOIT ATTEMPTS (Unusual Gas Usage) ===")
print(f"Found {len(outlier_txs)} transactions with unusual gas usage")

if outlier_txs:
    # Convert to DataFrame for easier handling
    outliers_df = pd.DataFrame(outlier_txs)
    outliers_df = outliers_df.sort_values('deviation', ascending=False)

    print("\nTop unusual transactions by gas usage deviation:")
    for _, row in outliers_df.head(10).iterrows():
        error_status = "Failed" if not pd.isna(row['error_code']) else "Success"
        print(f"TX: {row['tx_hash']} | Method: {row['method']} | "
              f"Gas: {row['gas_fee']:.8f} ETH | Normal: {row['method_avg_gas']:.8f} ETH | "
              f"Deviation: {row['deviation']:.2f}σ | Status: {error_status}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import timedelta

def analyze_failed_transactions_clusters(df, time_window_seconds=300):
    """
    Analyzes and clusters failed transactions to identify patterns.

    Parameters:
    df (pandas.DataFrame): DataFrame containing Ethereum transaction data
    time_window_seconds (int): Time window in seconds to consider transactions as part of same cluster

    Returns:
    pandas.DataFrame: Processed dataframe with cluster information
    """
    # Filter for failed transactions
    if 'ErrCode' not in df.columns:
        print("No ErrCode column found in dataframe. Cannot analyze failed transactions.")
        return None

    failed_txs = df[~df['ErrCode'].isna()].copy()

    print(f"=== FAILED TRANSACTION CLUSTER ANALYSIS ===")
    print(f"Analyzing {len(failed_txs)} failed transactions")

    if len(failed_txs) == 0:
        print("No failed transactions found in the dataset.")
        return None

    # 1. Error code distribution
    print("\n1. Error code distribution:")
    error_counts = failed_txs['ErrCode'].value_counts()
    for error, count in error_counts.items():
        percentage = count / len(failed_txs) * 100
        print(f"{error}: {count} txs ({percentage:.2f}%)")

    # 2. Time-based clustering of failed transactions
    print("\n2. Time clustering of failed transactions:")

    # Sort by timestamp
    failed_txs_sorted = failed_txs.sort_values('UnixTimestamp')

    # Calculate time differences between consecutive failed transactions
    if len(failed_txs_sorted) > 1:
        failed_txs_sorted['prev_timestamp'] = failed_txs_sorted['UnixTimestamp'].shift(1)
        failed_txs_sorted['time_diff'] = failed_txs_sorted['UnixTimestamp'] - failed_txs_sorted['prev_timestamp']

        # Define what constitutes a "cluster" (e.g., transactions within time_window_seconds of each other)
        cluster_threshold = time_window_seconds

        # Identify the start of each cluster
        failed_txs_sorted['new_cluster'] = failed_txs_sorted['time_diff'] > cluster_threshold
        failed_txs_sorted['cluster_id'] = failed_txs_sorted['new_cluster'].cumsum()

        # Analyze clusters
        clusters = failed_txs_sorted.groupby('cluster_id')

        print(f"Found {len(clusters)} distinct clusters of failed transactions")

        # Print info about the largest clusters
        cluster_sizes = failed_txs_sorted['cluster_id'].value_counts()
        large_clusters = cluster_sizes[cluster_sizes > 1].sort_values(ascending=False)

        if len(large_clusters) > 0:
            print(f"\nLargest clusters of failed transactions:")
            for cluster_id, size in large_clusters.head(5).items():
                cluster_txs = failed_txs_sorted[failed_txs_sorted['cluster_id'] == cluster_id]
                start_time = pd.to_datetime(cluster_txs['Datetime'].iloc[0])
                end_time = pd.to_datetime(cluster_txs['Datetime'].iloc[-1])
                duration = (end_time - start_time).total_seconds() / 60

                # Get unique error codes, methods and addresses in this cluster
                cluster_errors = cluster_txs['ErrCode'].value_counts()
                cluster_methods = cluster_txs['Method'].value_counts()
                cluster_senders = cluster_txs['From'].nunique()
                cluster_recipients = cluster_txs['To'].nunique()

                print(f"Cluster {cluster_id}: {size} failed transactions over {duration:.1f} minutes")
                print(f"  Time range: {start_time} to {end_time}")
                print(f"  Error codes: {', '.join(cluster_errors.index.tolist())}")
                print(f"  Methods: {', '.join(cluster_methods.index.tolist())}")
                print(f"  Unique senders: {cluster_senders}, Unique recipients: {cluster_recipients}")

                # Check if there are common senders or recipients
                if cluster_senders < 5:  # Only show if there are few senders
                    common_senders = cluster_txs['From'].value_counts().head(3)
                    sender_percentages = (common_senders / size * 100).round(1)
                    common_senders_str = ", ".join([f"{addr} ({count} txs, {pct}%)"
                                                 for (addr, count), pct in zip(common_senders.items(), sender_percentages)])
                    print(f"  Common senders: {common_senders_str}")

                if cluster_recipients < 5:  # Only show if there are few recipients
                    common_recipients = cluster_txs['To'].value_counts().head(3)
                    recipient_percentages = (common_recipients / size * 100).round(1)
                    common_recipients_str = ", ".join([f"{addr} ({count} txs, {pct}%)"
                                                    for (addr, count), pct in zip(common_recipients.items(), recipient_percentages)])
                    print(f"  Common recipients: {common_recipients_str}")

                print(f"  Example tx hash: {cluster_txs['Transaction Hash'].iloc[0]}")
                print("  ---")

        # 3. Error code-specific clusters
        print("\n3. Error code-specific clusters:")

        for error_code in error_counts.index[:5]:  # Analyze top 5 error codes
            error_txs = failed_txs_sorted[failed_txs_sorted['ErrCode'] == error_code]

            print(f"\nAnalyzing clusters for error code: {error_code} ({len(error_txs)} transactions)")

            # Apply clustering to these specific error transactions
            if len(error_txs) > 1:
                error_txs = error_txs.sort_values('UnixTimestamp')
                error_txs['prev_timestamp'] = error_txs['UnixTimestamp'].shift(1)
                error_txs['time_diff'] = error_txs['UnixTimestamp'] - error_txs['prev_timestamp']
                error_txs['new_cluster'] = error_txs['time_diff'] > cluster_threshold
                error_txs['cluster_id'] = error_txs['new_cluster'].cumsum()

                error_clusters = error_txs.groupby('cluster_id')
                error_cluster_sizes = error_txs['cluster_id'].value_counts()
                error_large_clusters = error_cluster_sizes[error_cluster_sizes > 1].sort_values(ascending=False)

                print(f"  Found {len(error_clusters)} clusters, with {len(error_large_clusters)} containing multiple transactions")

                if len(error_large_clusters) > 0:
                    largest_cluster_id = error_large_clusters.index[0]
                    largest_cluster = error_txs[error_txs['cluster_id'] == largest_cluster_id]

                    print(f"  Largest cluster has {len(largest_cluster)} transactions")
                    print(f"  Time span: {pd.to_datetime(largest_cluster['Datetime'].iloc[0])} to {pd.to_datetime(largest_cluster['Datetime'].iloc[-1])}")

                    # Check for method patterns
                    method_counts = largest_cluster['Method'].value_counts()
                    print(f"  Methods in largest cluster: {', '.join([f'{m} ({c})' for m, c in method_counts.items()])}")

                    # Check for address patterns
                    common_senders = largest_cluster['From'].value_counts().head(2)
                    common_recipients = largest_cluster['To'].value_counts().head(2)

                    if not common_senders.empty:
                        print(f"  Most common senders: {', '.join([f'{addr} ({count})' for addr, count in common_senders.items()])}")

                    if not common_recipients.empty:
                        print(f"  Most common recipients: {', '.join([f'{addr} ({count})' for addr, count in common_recipients.items()])}")



        # 4. Sender address failure analysis
        print("\n5. Sender address failure analysis:")

        # Get senders with the most failures
        sender_failures = failed_txs['From'].value_counts().head(5)

        for sender, failure_count in sender_failures.items():
            sender_txs = df[df['From'] == sender]
            total_sender_txs = len(sender_txs)
            failure_rate = failure_count / total_sender_txs * 100

            print(f"\nSender {sender}: {failure_count}/{total_sender_txs} failed transactions ({failure_rate:.2f}% failure rate)")

            # Error code distribution for this sender
            sender_errors = failed_txs[failed_txs['From'] == sender]['ErrCode'].value_counts()
            sender_error_str = ", ".join([f"{err} ({count})" for err, count in sender_errors.items()])
            print(f"  Error codes: {sender_error_str}")

            # Check if failures are clustered in time
            sender_failed_txs = failed_txs[failed_txs['From'] == sender].sort_values('UnixTimestamp')

            if len(sender_failed_txs) > 1:
                sender_failed_txs['time_diff'] = sender_failed_txs['UnixTimestamp'].diff()
                avg_time_between = sender_failed_txs['time_diff'].mean()
                min_time_between = sender_failed_txs['time_diff'].min()

                if not np.isnan(avg_time_between):
                    print(f"  Average time between failures: {avg_time_between/60:.2f} minutes")
                    print(f"  Minimum time between failures: {min_time_between:.1f} seconds")

                    # Detect rapid sequences of failures (potential retry patterns)
                    rapid_retries = sender_failed_txs[sender_failed_txs['time_diff'] < 60]  # Less than 1 minute
                    if len(rapid_retries) > 0:
                        print(f"  Found {len(rapid_retries)} instances of rapid retries (< 1 minute)")

                # Check for nonce-related issues
                if 'Nonce' in sender_failed_txs.columns:
                    nonce_diffs = sender_failed_txs['Nonce'].diff()
                    consecutive_nonces = (nonce_diffs == 1).sum()
                    if consecutive_nonces > 0:
                        print(f"  Found {consecutive_nonces} instances of consecutive nonce failures (potential sequence failures)")

                # Show most common methods with failures
                sender_methods = sender_failed_txs['Method'].value_counts().head(3)
                sender_method_str = ", ".join([f"{method} ({count})" for method, count in sender_methods.items()])
                print(f"  Most common failed methods: {sender_method_str}")

    return failed_txs_sorted  # Return the processed dataframe with cluster information

# Define high gas threshold (e.g., top 5%)
high_gas_threshold = df['TxnFee(ETH)'].quantile(0.95)
high_gas_txs = df[df['TxnFee(ETH)'] > high_gas_threshold]

print(f"=== HIGH GAS TRANSACTION ANALYSIS ===")
print(f"Analyzing {len(high_gas_txs)} transactions with gas fees > {high_gas_threshold:.8f} ETH")

# 1. Method Analysis for High Gas Transactions
print("\n1. Methods used in high gas transactions:")
high_gas_methods = high_gas_txs['Method'].value_counts()
for method, count in high_gas_methods.items():
    percentage = count / len(high_gas_txs) * 100
    percentage_of_method = count / len(df[df['Method'] == method]) * 100
    print(f"{method}: {count} txs ({percentage:.2f}% of high gas txs, {percentage_of_method:.2f}% of all {method} calls)")

# 2. Address Analysis
print("\n2. Addresses involved in high gas transactions:")

# Senders (From addresses)
high_gas_senders = high_gas_txs['From'].value_counts()
print("\nTop senders of high gas transactions:")
for addr, count in high_gas_senders.head(10).items():
    percentage = count / len(high_gas_txs) * 100
    total_from_addr = len(df[df['From'] == addr])
    percentage_of_addr = count / total_from_addr * 100
    print(f"{addr}: {count} txs ({percentage:.2f}% of high gas txs, {percentage_of_addr:.2f}% of this address's txs)")

# Recipients (To addresses)
high_gas_recipients = high_gas_txs['To'].value_counts()
print("\nTop recipients of high gas transactions:")
for addr, count in high_gas_recipients.head(10).items():
    percentage = count / len(high_gas_txs) * 100
    total_to_addr = len(df[df['To'] == addr])
    percentage_of_addr = count / total_to_addr * 100
    print(f"{addr}: {count} txs ({percentage:.2f}% of high gas txs, {percentage_of_addr:.2f}% of txs to this address)")

# 3. Status and Error Analysis - IMPROVED
if 'ErrCode' in df.columns:
    print("\n3. Transaction status analysis for high gas transactions:")

    # Add a clear status column
    high_gas_txs_copy = high_gas_txs.copy()
    high_gas_txs_copy['Status'] = high_gas_txs_copy['ErrCode'].apply(lambda x: 'Failed' if not pd.isna(x) else 'Success')

    # Analyze status counts
    status_counts = high_gas_txs_copy['Status'].value_counts()
    for status, count in status_counts.items():
        percentage = count / len(high_gas_txs_copy) * 100
        print(f"{status}: {count} txs ({percentage:.2f}%)")

    # If transactions failed, show the error codes
    if 'Failed' in status_counts:
        failed_txs = high_gas_txs_copy[high_gas_txs_copy['Status'] == 'Failed']
        error_counts = failed_txs['ErrCode'].value_counts()

        print("\nError codes in failed high gas transactions:")
        for error, count in error_counts.items():
            percentage = count / len(failed_txs) * 100
            print(f"  {error}: {count} txs ({percentage:.2f}% of failed txs)")

    # Compare success/failure rates between high gas and normal gas transactions
    normal_gas_txs = df[df['TxnFee(ETH)'] <= high_gas_threshold].copy()
    normal_gas_txs['Status'] = normal_gas_txs['ErrCode'].apply(lambda x: 'Failed' if not pd.isna(x) else 'Success')
    normal_status_counts = normal_gas_txs['Status'].value_counts(normalize=True) * 100
    high_status_counts = high_gas_txs_copy['Status'].value_counts(normalize=True) * 100

    print("\nComparison of success/failure rates between high gas and normal gas transactions:")
    for status in ['Success', 'Failed']:
        high_rate = high_status_counts.get(status, 0)
        normal_rate = normal_status_counts.get(status, 0)
        difference = high_rate - normal_rate
        print(f"{status}: High gas: {high_rate:.2f}%, Normal gas: {normal_rate:.2f}%, Difference: {difference:.2f}%")

# 4. Time Clustering Analysis
print("\n4. Time clustering of high gas transactions:")

# Sort by timestamp
high_gas_txs_sorted = high_gas_txs.sort_values('UnixTimestamp')

# Calculate time differences between consecutive high gas transactions
if len(high_gas_txs_sorted) > 1:
    high_gas_txs_sorted['prev_timestamp'] = high_gas_txs_sorted['UnixTimestamp'].shift(1)
    high_gas_txs_sorted['time_diff'] = high_gas_txs_sorted['UnixTimestamp'] - high_gas_txs_sorted['prev_timestamp']

    # Define what constitutes a "cluster" (e.g., transactions within 5 minutes of each other)
    cluster_threshold = 300  # 5 minutes in seconds

    # Identify the start of each cluster
    high_gas_txs_sorted['new_cluster'] = high_gas_txs_sorted['time_diff'] > cluster_threshold
    high_gas_txs_sorted['cluster_id'] = high_gas_txs_sorted['new_cluster'].cumsum()

    # Analyze clusters
    clusters = high_gas_txs_sorted.groupby('cluster_id')

    print(f"Found {len(clusters)} distinct clusters of high gas transactions")

    # Print info about the largest clusters
    cluster_sizes = high_gas_txs_sorted['cluster_id'].value_counts()
    large_clusters = cluster_sizes[cluster_sizes > 1].sort_values(ascending=False)

    if len(large_clusters) > 0:
        print(f"\nLargest clusters of high gas transactions:")
        for cluster_id, size in large_clusters.head(5).items():
            cluster_txs = high_gas_txs_sorted[high_gas_txs_sorted['cluster_id'] == cluster_id]
            start_time = pd.to_datetime(cluster_txs['Datetime'].iloc[0])
            end_time = pd.to_datetime(cluster_txs['Datetime'].iloc[-1])
            duration = (end_time - start_time).total_seconds() / 60

            # Get unique methods and addresses in this cluster
            cluster_methods = cluster_txs['Method'].value_counts()
            cluster_senders = cluster_txs['From'].nunique()

            print(f"Cluster {cluster_id}: {size} transactions over {duration:.1f} minutes")
            print(f"  Time range: {start_time} to {end_time}")
            print(f"  Methods: {', '.join(cluster_methods.index.tolist())}")
            print(f"  Unique senders: {cluster_senders}")
            print(f"  Example tx hash: {cluster_txs['Transaction Hash'].iloc[0]}")

            # Calculate success rate for this cluster
            if 'ErrCode' in cluster_txs.columns:
                success_count = cluster_txs['ErrCode'].isna().sum()
                success_rate = (success_count / len(cluster_txs)) * 100
                print(f"  Success rate: {success_rate:.1f}% ({success_count}/{len(cluster_txs)} transactions)")

            print("  ---")

    # Identify bursts of activity (many transactions in short time)
    # Calculate transactions per minute during active periods
    cluster_durations = []
    for _, cluster in clusters:
        if len(cluster) > 1:
            start_time = pd.to_datetime(cluster['Datetime'].iloc[0])
            end_time = pd.to_datetime(cluster['Datetime'].iloc[-1])
            duration_minutes = max(1, (end_time - start_time).total_seconds() / 60)  # Minimum 1 minute
            tx_per_minute = len(cluster) / duration_minutes

            cluster_durations.append({
                'cluster_id': cluster['cluster_id'].iloc[0],
                'size': len(cluster),
                'duration_minutes': duration_minutes,
                'tx_per_minute': tx_per_minute,
                'start_time': start_time,
                'end_time': end_time
            })

    if cluster_durations:
        cluster_df = pd.DataFrame(cluster_durations)
        high_intensity_clusters = cluster_df.sort_values('tx_per_minute', ascending=False)

        print("\nHighest intensity clusters (transactions per minute):")
        for _, cluster in high_intensity_clusters.head(5).iterrows():
            print(f"Cluster {cluster['cluster_id']}: {cluster['size']} txs in {cluster['duration_minutes']:.1f} min "
                  f"({cluster['tx_per_minute']:.2f} tx/min)")
            print(f"  Time range: {cluster['start_time']} to {cluster['end_time']}")

            # Get more details about this specific cluster
            cluster_txs = high_gas_txs_sorted[high_gas_txs_sorted['cluster_id'] == cluster['cluster_id']]
            cluster_methods = cluster_txs['Method'].value_counts()
            common_senders = cluster_txs['From'].value_counts().head(3)

            print(f"  Methods: {', '.join(cluster_methods.index.tolist())}")
            print(f"  Top senders: {', '.join(common_senders.index.tolist())}")

            # Add success/failure analysis for this cluster
            if 'ErrCode' in cluster_txs.columns:
                success_count = cluster_txs['ErrCode'].isna().sum()
                failure_count = len(cluster_txs) - success_count
                print(f"  Transaction status: {success_count} successful, {failure_count} failed")

            print("  ---")

# 5. Full details of the highest gas and failed high gas transactions
print("\n6. Details of top 5 highest gas transactions:")
top_gas_txs = df.sort_values('TxnFee(ETH)', ascending=False).head(5)
for i, (_, row) in enumerate(top_gas_txs.iterrows(), 1):
    status = 'Success' if pd.isna(row.get('ErrCode')) else f"Failed: {row['ErrCode']}"
    print(f"\nTransaction #{i} - Gas: {row['TxnFee(ETH)']:.8f} ETH - Status: {status}")
    print(f"Hash: {row['Transaction Hash']}")
    print(f"From: {row['From']}")
    print(f"To: {row['To']}")
    print(f"Method: {row['Method']}")
    print(f"Value IN: {row['Value_IN(ETH)']} ETH")
    print(f"Value OUT: {row['Value_OUT(ETH)']} ETH")
    print(f"Time: {row['Datetime']}")

# Also show details of failed high gas transactions
if 'ErrCode' in df.columns:
    failed_high_gas = high_gas_txs[~high_gas_txs['ErrCode'].isna()].sort_values('TxnFee(ETH)', ascending=False)

    if len(failed_high_gas) > 0:
        print("\nDetails of top 5 failed high gas transactions:")
        for i, (_, row) in enumerate(failed_high_gas.head(5).iterrows(), 1):
            print(f"\nFailed Transaction #{i} - Gas: {row['TxnFee(ETH)']:.8f} ETH - Error: {row['ErrCode']}")
            print(f"Hash: {row['Transaction Hash']}")
            print(f"From: {row['From']}")
            print(f"To: {row['To']}")
            print(f"Method: {row['Method']}")
            print(f"Value IN: {row['Value_IN(ETH)']} ETH")
            print(f"Value OUT: {row['Value_OUT(ETH)']} ETH")
            print(f"Time: {row['Datetime']}")

# Run the failed transaction cluster analysis
print("\n=== RUNNING FAILED TRANSACTION ANALYSIS ===")
failed_clusters = analyze_failed_transactions_clusters(df)
```

```python
def analyze_hacker_transactions(df, hacker_addresses):
    """
    Filter and analyze transactions from known hacker addresses

    Parameters:
    df (pd.DataFrame): DataFrame containing blockchain transaction data
    hacker_addresses (list): List of known hacker addresses to analyze

    Returns:
    pd.DataFrame: Filtered DataFrame containing only transactions from hacker addresses
    """
    # Convert addresses to lowercase for case-insensitive comparison
    hacker_addresses = [addr.lower() for addr in hacker_addresses]

    # Filter transactions where From address is in the hacker_addresses list
    hacker_txs = df[df['From'].str.lower().isin(hacker_addresses)].copy()

    # Print summary statistics
    print(f"Found {len(hacker_txs)} transactions from known hacker addresses")

    if len(hacker_txs) == 0:
        print("No transactions found from the specified hacker addresses.")
        return hacker_txs

    # Add a column to identify which hacker address the transaction is from
    hacker_txs['HackerID'] = hacker_txs['From'].str.lower().apply(lambda x: hacker_addresses.index(x))

    # Create visualizations to analyze hacker activity
    visualize_hacker_activity(hacker_txs, hacker_addresses)

    return hacker_txs

def visualize_hacker_activity(hacker_txs, hacker_addresses):
    """
    Create visualizations to analyze hacker activity patterns

    Parameters:
    hacker_txs (pd.DataFrame): DataFrame containing filtered hacker transactions
    hacker_addresses (list): List of hacker addresses being analyzed
    """
    # Set up the plotting style
    plt.style.use('ggplot')
    sns.set(style="darkgrid")

    # Create a figure with multiple subplots
    fig = plt.figure(figsize=(20, 16))

    # 1. Transaction Timeline - when are hackers most active?
    ax1 = plt.subplot(3, 2, 1)

    # Convert DateTime to datetime objects if it's not already
    if not pd.api.types.is_datetime64_any_dtype(hacker_txs['DateTime (UTC)']):
        hacker_txs['DateTime (UTC)'] = pd.to_datetime(hacker_txs['DateTime (UTC)'])

    # Group by date and count transactions
    daily_counts = hacker_txs.groupby(hacker_txs['DateTime (UTC)'].dt.date).size()
    daily_counts.plot(kind='line', ax=ax1, marker='o', linewidth=2)
    ax1.set_title("Hacker Activity Timeline", fontsize=14)
    ax1.set_xlabel("Date", fontsize=12)
    ax1.set_ylabel("Number of Transactions", fontsize=12)

    # 2. Transaction Value Distribution - what size transactions are they making?
    ax2 = plt.subplot(3, 2, 2)

    # Combine inbound and outbound values
    value_columns = ['Value_IN(ETH)', 'Value_OUT(ETH)']
    for col in value_columns:
        if col in hacker_txs.columns:
            sns.histplot(hacker_txs[col].dropna(), ax=ax2, label=col, kde=True, alpha=0.6)

    ax2.set_title("Transaction Value Distribution", fontsize=14)
    ax2.set_xlabel("Value (ETH)", fontsize=12)
    ax2.set_ylabel("Frequency", fontsize=12)
    ax2.legend()

    # 3. Transaction Methods Used - what contract methods are they calling?
    ax3 = plt.subplot(3, 2, 3)

    if 'Method' in hacker_txs.columns:
        method_counts = hacker_txs['Method'].value_counts().head(10)
        method_counts.plot(kind='barh', ax=ax3, color='teal')
        ax3.set_title("Top Methods Used by Hackers", fontsize=14)
        ax3.set_xlabel("Number of Transactions", fontsize=12)
        ax3.set_ylabel("Method", fontsize=12)

    # 4. Destination Addresses - where are they sending funds?
    ax4 = plt.subplot(3, 2, 4)

    top_destinations = hacker_txs['To'].value_counts().head(10)
    top_destinations.plot(kind='barh', ax=ax4, color='purple')
    ax4.set_title("Top Destination Addresses", fontsize=14)
    ax4.set_xlabel("Number of Transactions", fontsize=12)
    ax4.set_ylabel("Destination Address", fontsize=12)

    # 5. Success vs Failed Transactions
    ax5 = plt.subplot(3, 2, 5)

    if 'Status' in hacker_txs.columns:
        status_counts = hacker_txs['Status'].value_counts()
        status_counts.plot(kind='pie', ax=ax5, autopct='%1.1f%%', startangle=90,
                          colors=['green', 'red'] if len(status_counts) > 1 else ['green'])
        ax5.set_title("Transaction Status Distribution", fontsize=14)
        ax5.set_ylabel("")

    # 6. Gas Fees Paid Over Time - are they paying premium gas fees?
    ax6 = plt.subplot(3, 2, 6)

    if 'TxnFee(ETH)' in hacker_txs.columns:
        # Plot gas fees over time
        plt.scatter(hacker_txs['DateTime (UTC)'], hacker_txs['TxnFee(ETH)'],
                   c=hacker_txs['HackerID'], cmap='viridis', alpha=0.7)
        ax6.set_title("Gas Fees Paid Over Time", fontsize=14)
        ax6.set_xlabel("Date", fontsize=12)
        ax6.set_ylabel("Transaction Fee (ETH)", fontsize=12)

    # Add a color-coded legend for hacker addresses
    handles = []
    for i, addr in enumerate(hacker_addresses):
        handles.append(plt.Line2D([0], [0], marker='o', color='w',
                                  markerfacecolor=plt.cm.viridis(i/len(hacker_addresses)),
                                  markersize=10, label=f"{addr[:10]}..."))
    ax6.legend(handles=handles, title="Hacker Addresses", loc='upper right')

    plt.tight_layout()
    plt.savefig('hacker_activity_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()


hacker_addresses = [
    "0xdd18f47b3d4ead87acad319871979370419e227d",
    "0x190aa2f700c606c063ecea034016721a7b97529a"
]

# Assuming df is your DataFrame with transaction data
hacker_transactions = analyze_hacker_transactions(df, hacker_addresses)

# Print detailed information about large value transactions
large_value_txs = hacker_transactions[hacker_transactions['Value_OUT(ETH)'] > 1.0]
print(f"\nFound {len(large_value_txs)} large value outbound transactions:")
for idx, tx in large_value_txs.iterrows():
    print(f"Tx: {tx['Transaction Hash']} | From: {tx['From']} | To: {tx['To']} | Value: {tx['Value_OUT(ETH)']:.4f} ETH | Date: {tx['DateTime (UTC)']}")
```

```python
from collections import Counter

def identify_related_hackers(df, known_hacker_addresses, target_date="2025-02-18"):
    """
    Identify potential additional hackers based on transactions that:
    1. Occurred on the specified date
    2. Failed/reverted
    3. Come from different addresses than our known hackers

    Parameters:
    df (pd.DataFrame): DataFrame containing blockchain transaction data
    known_hacker_addresses (list): List of known hacker addresses
    target_date (str): Specific date to analyze in 'YYYY-MM-DD' format

    Returns:
    pd.DataFrame: Filtered DataFrame containing potential related hacker transactions
    """
    print(f"Analyzing transactions from {target_date}...")

    # Convert known hacker addresses to lowercase
    known_hacker_addresses = [addr.lower() for addr in known_hacker_addresses]

    # Filter by specific date
    if not pd.api.types.is_datetime64_any_dtype(df['DateTime (UTC)']):
        df['DateTime (UTC)'] = pd.to_datetime(df['DateTime (UTC)'])

    target_date = pd.to_datetime(target_date)
    date_txs = df[df['DateTime (UTC)'].dt.date == target_date.date()].copy()

    print(f"Found {len(date_txs)} transactions on {target_date.date()}")

    # Identify failed transactions
    failed_txs = pd.DataFrame()

    # Specifically check for "execution reverted" in ErrCode column
    if 'ErrCode' in date_txs.columns:
        failed_txs = date_txs[date_txs['ErrCode'].str.lower().str.contains('execution reverted', na=False)]
        if len(failed_txs) > 0:
            print(f"Found {len(failed_txs)} transactions with 'execution reverted' in ErrCode")

    # If not found, try other failure indicators
    if len(failed_txs) == 0 and 'Status' in date_txs.columns:
        failed_txs = date_txs[
            (date_txs['Status'].str.lower() == 'fail') |
            (date_txs['Status'].str.lower() == 'failed') |
            (date_txs['Status'].str.lower().str.contains('fail')) |
            (date_txs['Status'].str.lower().str.contains('revert'))
        ]

    # Try searching in all columns for "execution reverted"
    if len(failed_txs) == 0:
        for col in date_txs.columns:
            if date_txs[col].dtype == object:  # Only check string columns
                col_failed = date_txs[date_txs[col].str.lower().str.contains('execution reverted', na=False)]
                if len(col_failed) > 0:
                    failed_txs = col_failed
                    print(f"Found {len(failed_txs)} transactions with 'execution reverted' in {col}")
                    break

    if len(failed_txs) == 0:
        print("WARNING: Could not identify any failed transactions!")
        return pd.DataFrame(), {}

    print(f"Found {len(failed_txs)} failed transactions")

    # Exclude transactions from known hackers
    potential_related_txs = failed_txs[~failed_txs['From'].str.lower().isin(known_hacker_addresses)]
    print(f"Found {len(potential_related_txs)} potential related hacker transactions")

    # Get top 5 potential hacker addresses by transaction count
    if len(potential_related_txs) > 0:
        potential_hackers = potential_related_txs['From'].value_counts().head(5)

        print(f"\nTop 5 potential related hacker addresses:")
        for addr, count in potential_hackers.items():
            print(f"- {addr}: {count} suspicious transactions")

        # Print transaction hashes for top 5 hackers
        print_tx_hashes(potential_related_txs, potential_hackers.index.tolist())

    return potential_related_txs

def print_tx_hashes(df, addresses):
    """
    Print transaction hashes for specified addresses

    Parameters:
    df (pd.DataFrame): DataFrame containing transaction data
    addresses (list): List of addresses to print hashes for
    """
    print("\nTransaction hashes for top potential hackers:")

    for addr in addresses:
        addr_txs = df[df['From'] == addr]
        if 'Transaction Hash' in addr_txs.columns:
            tx_column = 'Transaction Hash'
        elif 'Txhash' in addr_txs.columns:
            tx_column = 'Txhash'
        elif 'Hash' in addr_txs.columns:
            tx_column = 'Hash'
        else:
            # Try to find any column that might contain transaction hashes
            hash_columns = [col for col in addr_txs.columns if 'hash' in col.lower() or 'tx' in col.lower()]
            if hash_columns:
                tx_column = hash_columns[0]
            else:
                print(f"No transaction hash column found for address {addr}")
                continue

        print(f"\n{addr}:")
        for tx_hash in addr_txs[tx_column].values:
            print(f"  {tx_hash}")


known_hacker_addresses = [
    "0xdd18f47b3d4ead87acad319871979370419e227d",
    "0x190aa2f700c606c063ecea034016721a7b97529a"
]

# Call the function with the specific date
potential_hackers_txs = identify_related_hackers(df, known_hacker_addresses, target_date="2025-02-18")
```

```python
def print_suspicious_transactions(df, suspicious_addresses):
    """
    Filter DataFrame for specific suspicious addresses and print address, time, and transaction hash

    Parameters:
    df (pd.DataFrame): DataFrame containing blockchain transaction data
    suspicious_addresses (list): List of suspicious addresses to filter for
    """
    # Ensure datetime column is properly formatted
    if not pd.api.types.is_datetime64_any_dtype(df['DateTime (UTC)']):
        df['DateTime (UTC)'] = pd.to_datetime(df['DateTime (UTC)'])

    # Identify transaction hash column
    hash_column = None
    for possible_name in ['Transaction Hash', 'Txhash', 'Hash']:
        if possible_name in df.columns:
            hash_column = possible_name
            break

    if not hash_column:
        hash_columns = [col for col in df.columns if 'hash' in col.lower() or 'tx' in col.lower()]
        if hash_columns:
            hash_column = hash_columns[0]
        else:
            print("Error: No transaction hash column found")
            return

    print("\n=== SUSPICIOUS ADDRESSES TRANSACTION DETAILS ===")

    for address in suspicious_addresses:
        # Filter for this address
        address_txs = df[df['From'].str.lower() == address.lower()]

        if len(address_txs) == 0:
            print(f"\nNo transactions found for address: {address}")
            continue

        print(f"\nAddress: {address} - {len(address_txs)} transactions")

        # Sort by datetime
        address_txs = address_txs.sort_values('DateTime (UTC)')

        # Print details for each transaction
        for _, tx in address_txs.iterrows():
            timestamp = tx['DateTime (UTC)'].strftime('%Y-%m-%d %H:%M:%S')
            tx_hash = tx[hash_column]

            # Check if transaction is failed
            is_failed = False
            if 'ErrCode' in tx and pd.notna(tx['ErrCode']):
                if isinstance(tx['ErrCode'], str) and 'execution reverted' in tx['ErrCode'].lower():
                    is_failed = True

            status = "FAILED" if is_failed else "SUCCESS"
            print(f"  {timestamp} | {tx_hash} | {status}")

        # Print summary
        failed_txs = address_txs[address_txs['ErrCode'].str.lower().str.contains('execution reverted', na=False)]
        failure_rate = len(failed_txs) / len(address_txs) * 100
        print(f"  Summary: {len(failed_txs)}/{len(address_txs)} failed ({failure_rate:.2f}%)")

suspicious_addresses = [
    "0x6c70dd1293edb55e8748828174f5a12b6256921a",
    "0x88b2f36284b79978a87b5480465c42151aa6459a",
    "0xe9a9d424fc352a91e2a834b1b27995ee13c17343",
    "0x04b11a7fe79c84f0891ff3d6fe93f3511c3e33ad",
    "0xeeb17adefd06a3c67abd4adf7e42978b1f018c8e"
]

# Call the function
print_suspicious_transactions(df, suspicious_addresses)
```
